#### Q. You want to build a DECISION TREE model predicting `isBar` (whether or not a coupon was used at a bar) based on `ageActual`, `time`, `temperature`, `gender`, and `maritalStatus`. Create the appropriate `outcome` and `features` variable for this model using the `df2` dataframe.

```python
outcome = df2["isBar"]
outcome

numericFeatures = df2[["ageActual", "temperature"]]
dummies = pd.get_dummies(df2[["time", "gender", "maritalStatus"]], drop_first = True)
features = pd.concat([numericFeatures, dummies], axis = 1)
features
```

#### Q. Partition the data so that 25% of the rows are used to evaluate the strength of the model on "new data" and use a random state of 67.

```python
from sklearn.model_selection import train_test_split
featuresTrain, featuresTest, outcomeTrain, outcomeTest = train_test_split(features, outcome,
                                                                         test_size = 0.25,
                                                                         random_state = 67)
```

#### Q. Build a Decision Tree model to predict your `outcome` variable from the `features` using a random state of 67. Be sure to provide code that allows you to assess the strength of the model and whether there is overfitting.

#### Note: Because your `outcome` variable is `True`/`False` and not a string, the code to display your tree with the `plot_tree()` command will not work. For the purposes of this question, you do NOT need to display your tree.

```python
import sklearn.tree
modTree = sklearn.tree.DecisionTreeClassifier(random_state = 67)
modTree.fit(featuresTrain, outcomeTrain)

predTreeTrain = modTree.predict(featuresTrain)
predTreeTest = modTree.predict(featuresTest)

print(sklearn.metrics.classification_report(outcomeTrain, predTreeTrain))
cmTrain = sklearn.metrics.confusion_matrix(outcomeTrain, predTreeTrain)
sklearn.metrics.ConfusionMatrixDisplay(cmTrain, display_labels = modTree.classes_).plot()

print(sklearn.metrics.classification_report(outcomeTest, predTreeTest))
cmTest = sklearn.metrics.confusion_matrix(outcomeTest, predTreeTest)
sklearn.metrics.ConfusionMatrixDisplay(cmTest, display_labels = modTree.classes_).plot()
```

#### Q. How good is your model at predicting *new data*?

*The model is 81% accurate at predicting new data (based on the accuracy % for the test data.*

#### Q. Let's say you wanted a model that specifically predicts when someone will use a bar coupon. Do you have any concerns about this model's ability to predict this outcome? Why or why not?

*There doesn't seem to be a huge overfitting problem since the difference in accuracy between training and test is only 7%. However, the model has a higher f1 score for the `False` category (meaning the person did not use a bar coupon) -- 0.89 for `False` vs. 0.16 for `True`. This means that the model is much better at predicting when someone won't use a bar coupon, which is not what we want to predict. It's pretty bad at predicting when someone WILL use a bar coupon.*

*Therefore, I don't think this is a great model at predicting bar coupon usage.*
